```python
import pandas as pd
```

### Импорт датасета и создание датафрейма

```python
# Датасет отзывов на Яндекс Картах https://github.com/yandex/geo-reviews-dataset-2023.git
# Путь к файлу
file_path = '/home/jupyter/datasphere/project/final_project/geo-reviews-dataset-2023.tskv'

# Создание списка для хранения распарсенных строк
parsed_rows = []

# Открытие файла и преобразование датасета к нормальной таблице 
with open(file_path, 'r', encoding='utf8') as file:
    for line in file:
        row_dict = {}
        elements = line.strip().split('\t')
        for element in elements:
            if '=' in element:
                key, value = element.split('=', 1)
                row_dict[key] = value
        parsed_rows.append(row_dict)

# Создание датафрейма
df = pd.DataFrame(parsed_rows)
df
```

### Формирмирование данных для анализа отзывов на гостиницы

```python
# Разделим строки на отдельные рубрики
df['rubrics'] = df['rubrics'].str.split(';')
df_split_rubrics = df.explode('rubrics')

df_split_rubrics
```

```python
# Подсчитаем количество отзывов в каждой рубрике и выведем TOP 10 рубрик
df_split_rubrics['rubrics'].value_counts().head(10)
```

```python
# Выберем отзывы c рубрикой "Гостиница"
df_hotels_reviews = df_split_rubrics.query("rubrics == 'Гостиница'")
df_hotels_reviews
```

```python
# Удалим колонки rubrics, address и name_ru из датафрейма, так как они нам не понадобятся для анализа
df_hotels_reviews = df_hotels_reviews.drop(columns=['rubrics', 'address', 'name_ru'])
df_hotels_reviews
```

```python
# Проверим типы данных в датафрейме
df_hotels_reviews.info()
```

```python
# Преобразуем тип колонки rating из строки в целое число
df_hotels_reviews['rating'] = df_hotels_reviews['rating'].astype(float).astype(int)
df_hotels_reviews
```

```python
# Подсчитаем количество отзывов в разрезе рейтинга
df_hotels_reviews['rating'].value_counts()
```

```python
# Удалим отзывы с рейтингом = 0
df_hotels_reviews = df_hotels_reviews[df_hotels_reviews['rating'] != 0]
df_hotels_reviews['rating'].value_counts()
```

```python
# Подсчитаем количество отзывов, где количество символов меньше 10
short = (df_hotels_reviews['text'].str.len() < 10).sum()
print(f"Количество отзывов, где количество символов меньше 10: {short}")
```

```python
# Удалим отзывы, где количество символов меньше 10
df_hotels_reviews = df_hotels_reviews[df_hotels_reviews['text'].str.len() > 10]
```

```python
# Провериим датафрейм на наличие пропущенных значений в каждом столбце
print("Количество пропущенных значений в каждом столбце:")
df_hotels_reviews.isnull().sum()
```

```python
# Проверим датафрейм на дубликаты строк
duplicates = df_hotels_reviews.duplicated().sum()
print(f"Количество дублирующихся строк: {duplicates}")
```

```python
# Удалим дубликаты строк
df_hotels_reviews = df_hotels_reviews.drop_duplicates()
```
